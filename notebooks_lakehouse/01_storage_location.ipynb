### Criação do Storage Location

```python
%sql
CREATE EXTERNAL LOCATION training 
URL 'abfss://treinamentos@<storage-account>.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL main)
COMMENT "Usado para fins de projetos e treinamentos"
```

### Criação do schemas usando Medallion Architecture

```python
%sql

CREATE SCHEMA training.bronze MANAGED
LOCATION 'abfss://treinamentos@<storage-account>.dfs.core.windows.net/ipca_geral/bronze';
    
CREATE SCHEMA training.silver MANAGED
LOCATION 'abfss://treinamentos@<storage-account>.dfs.core.windows.net/ipca_geral/silver';
    
CREATE SCHEMA training.gold MANAGED
LOCATION 'abfss://treinamentos@<storage-account>.dfs.core.windows.net/ipca_geral/gold';
```
