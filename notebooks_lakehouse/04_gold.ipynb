```python
%sql
-- Coletar o preço medio por data_referencia
SELECT 
    data_referencia,
    ROUND(MIN(preco_fechamento), 2) AS preco_fechamento_min,
    ROUND(AVG(preco_fechamento), 2) AS preco_fechamento_media,
    ROUND(MAX(preco_fechamento), 2) AS preco_fechamento_max
FROM 
    training.silver.precos_acoes_b3
GROUP BY data_referencia
ORDER BY data_referencia
```
